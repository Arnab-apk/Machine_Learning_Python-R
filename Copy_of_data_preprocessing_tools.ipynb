# Data Preprocessing Tools

## Importing the libraries

```
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```
dataset =pd.read_csv('Data.csv')
#features are in generally the first column of the data set and the dependent variables are generally in the last column of the data set.
#matrix variable features
#: means range we have to take all the rows
#-1 means the last column
#iloc collects the index of the rows and colums we want
X=dataset.iloc[:,:-1].values
#dependent variable vector
y=dataset.iloc[:,-1].values
```

```
print(X)
```

```
print(y)
```

## Taking care of missing data

```
from sklearn.impute import SimpleImputer
#average salary
imputer=SimpleImputer(missing_values=np.nan,strategy='mean')
imputer.fit(X[:,1:3])
X[:,1:3]=imputer.transform(X[:,1:3])
```

```
print(X)
```

## Encoding categorical data

```
#encoding the names of the country in special code of 0 and 1
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct=ColumnTransformer(transformers=[('encoder',OneHotEncoder(),[0])],remainder='passthrough')
X=np.array(ct.fit_transform(X))
```

```
print(X)
```

# Data Preprocessing Tools

## Importing the libraries

```
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```
dataset =pd.read_csv('Data.csv')
#features are in generally the first column of the data set and the dependent variables are generally in the last column of the data set.
#matrix variable features
#: means range we have to take all the rows
#-1 means the last column
#iloc collects the index of the rows and colums we want
X=dataset.iloc[:,:-1].values#x for martrix of features
#dependent variable vector
y=dataset.iloc[:,-1].values#y for dependent variable vector
```

```
print(X)
```

```
print(y)
```

## Taking care of missing data

```
from sklearn.impute import SimpleImputer
#average salary
imputer=SimpleImputer(missing_values=np.nan,strategy='mean')
imputer.fit(X[:,1:3])
X[:,1:3]=imputer.transform(X[:,1:3])
```

```
print(X)
```

## Encoding categorical data

```
from sklearn.impute import SimpleImputer
imputer=SimpleImputer(missing_values=np.nan,strategy='mean')
#The imputer function finds the missing values and connects to teh matrix of features
imputer.fit(X[:,1:3])
#Updating the matrix of features X, the transform method returns teh updated matrix of features
X[:,1:3]=imputer.transform(X[:,1:3])
```

```
print(X)
```

# Encoding the catagorical data

### Encoding the Independent Variable

```
#turning the strings into numbers
#numerical order to train the model
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
#the transformer function takes 3 arguments the transformation ,transformation type and the column to encode
#the remainder(passthrough) keeps the numbers unchanged except than the encoded columns
ct=ColumnTransformer(transformers=[('encoder',OneHotEncoder(),[0])],remainder='passthrough')
X=np.array(ct.fit_transform(X))
print(X)
#the country names are encoded as Vectors:
#France-1.0.0
#Spain-0.0.1
#Germany-0.1.0
```
